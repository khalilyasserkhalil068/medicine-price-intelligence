# 04 - Generate Mappings

Generate key mapping CSVs for name-to-ID relationships.
These mappings will be used during the SQL loading phase to convert name-based foreign keys into integer ID-based foreign keys.

```python
import pandas as pd
from pathlib import Path
```

```python

# Paths
ROOT = Path().resolve().parent
CLEANED = ROOT / "data" / "cleaned"
MAPPINGS = ROOT / "reports" / "key_mappings"

MAPPINGS.mkdir(parents=True, exist_ok=True)
```

```python
# Generic mapping (unique names — clean 1:1 mapping)
gen = pd.read_csv(CLEANED / "generic_clean.csv")
gen_map = gen[["generic id", "generic name"]].drop_duplicates().sort_values("generic id")
gen_map.to_csv(MAPPINGS / "generic_mapping.csv", index=False)
print(f"generic_mapping.csv: {len(gen_map)} rows")
```

```python
# Manufacturer mapping (unique names — clean 1:1 mapping)
mfr = pd.read_csv(CLEANED / "manufacturer_clean.csv")
mfr_map = mfr[["manufacturer id", "manufacturer name"]].drop_duplicates().sort_values("manufacturer id")
mfr_map.to_csv(MAPPINGS / "manufacturer_mapping.csv", index=False)
print(f"manufacturer_mapping.csv: {len(mfr_map)} rows")
```

```python
# Dosage form mapping (unique names — clean 1:1 mapping)
dos = pd.read_csv(CLEANED / "dosage_form_clean.csv")
dos_map = dos[["dosage form id", "dosage form name"]].drop_duplicates().sort_values("dosage form id")
dos_map.to_csv(MAPPINGS / "dosage_form_mapping.csv", index=False)
print(f"dosage_form_mapping.csv: {len(dos_map)} rows")
```

```python
# Drug class mapping (DUPLICATES EXIST — use lowest ID per name as canonical)
dcl = pd.read_csv(CLEANED / "drug_class_clean.csv")
dup_names = dcl[dcl.duplicated(subset=["drug class name"], keep=False)]
print(f"\ndrug_class duplicates: {len(dup_names)} rows across {dup_names['drug class name'].nunique()} names")

dcl_map = dcl.groupby("drug class name")["drug class id"].min().reset_index()
dcl_map = dcl_map.rename(columns={"drug class id": "drug_class_id_canonical"})
dcl_map.columns = ["drug class name", "drug class id"]
dcl_map = dcl_map[["drug class id", "drug class name"]].sort_values("drug class id")
dcl_map.to_csv(MAPPINGS / "drug_class_mapping.csv", index=False)
print(f"drug_class_mapping.csv: {len(dcl_map)} rows (deduplicated, canonical lowest ID per name)")
```

```python
# Indication mapping (unique names — clean 1:1 mapping)
ind = pd.read_csv(CLEANED / "indication_clean.csv")
ind_map = ind[["indication id", "indication name"]].drop_duplicates().sort_values("indication id")
ind_map.to_csv(MAPPINGS / "indication_mapping.csv", index=False)
print(f"indication_mapping.csv: {len(ind_map)} rows")
```

```python
print("\nAll mapping tables saved to reports/key_mappings/")
```
